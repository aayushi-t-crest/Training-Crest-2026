```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
```

```python
chat = ChatGroq(
    model="llama-3.1-8b-instant",
    seed=365,
    temperature=0,
    max_tokens=100,
)
```

## Create A Message With Format Instructions

```python
parser = DatetimeOutputParser()

message_h = HumanMessage(
    content=f"""When was the Danish poet Piet Hein born?
{parser.get_format_instructions()}
"""
)
```

```python
print(message_h.content)
```

## Invoke The Model

```python
response = chat.invoke([message_h])
print(response.content)
```

## Parse The Model Output Into A Python `datetime`

```python
from pydantic import BaseModel, Field

# Define structured schema
class DateResponse(BaseModel):
    date: str = Field(description="The date in YYYY-MM-DD format")

chat = ChatGroq(
    model_name="llama-3.1-8b-instant",
    temperature=0
)

structured_llm = chat.with_structured_output(DateResponse)

response = structured_llm.invoke(
    "What date is 5 days from today? Return only the date in YYYY-MM-DD format."
)

print(response.date)
```
