## Exercise 1

**Follow along with these steps:**
* **Create a figure object called fig using plt.figure()**
* **Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax.**
* **Plot (x,y) on that axes and set the labels and titles to match the plot below:**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 100, 100)
y = 2 * x

fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.6, 0.6])  # smaller plot inside the figure

ax.plot(x, y)
ax.set_title("title")
ax.set_xlabel("x")
ax.set_ylabel("y")

plt.show()
```

## Exercise 3

**Create the plot below by adding two axes to a figure object at [0,0,1,1] and [0.2,0.5,.4,.4]**

```python
# Create the figure
fig = plt.figure()

# Create the first axes (full figure)
ax1 = fig.add_axes([0, 0, 1, 1])

# Create the second axes (inset)
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])

# Show the plot
plt.show()
```

**Now use x,y, and z arrays to recreate the plot below. Notice the xlimits and y limits on the inserted plot:**

```python
# Data
x = np.linspace(0, 100, 100)
z = x**2

# Create figure and axes
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6, 8))

# Plot on the first axes
ax1.plot(x, z, 'b')
ax1.set_xlabel('X')
ax1.set_ylabel('Z')

# Insert axes in the first plot
axins1 = ax1.inset_axes([0.2, 0.5, 0.4, 0.4])
axins1.plot(x, z, 'b')
axins1.set_xlim(20, 22)
axins1.set_ylim(30, 50)
axins1.set_title('zoom')

# Plot on the second axes
ax2.plot(x, z, 'b')
ax2.set_xlabel('X')
ax2.set_ylabel('Z')

# Inset axes in the second plot
axins2 = ax2.inset_axes([0.2, 0.5, 0.4, 0.4])
axins2.plot(x, z, 'b')
axins2.set_xlim(20, 22)
axins2.set_ylim(30, 50)
axins2.set_title('zoom')

# Show the plot
plt.tight_layout()
plt.show()
```

## Exercise 4

**Use plt.subplots(nrows=1, ncols=2) to create the plot below.**

```python
# Data
x = np.linspace(0, 100, 100)
y = x
z = x**2

# Create subplots with 1 row and 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Plot x vs y on the first axis with blue dashed line
ax1.plot(x, y, 'b--', linewidth=2)
ax1.set_title('x vs y')

# Plot x vs z on the second axis with red solid line
ax2.plot(x, z, 'r-', linewidth=2)
ax2.set_title('x vs z')

# Show the plot
plt.tight_layout()
plt.show()
```
